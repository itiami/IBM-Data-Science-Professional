### Task 3




##### Display the total payload mass carried by boosters launched by NASA (CRS)


```python
# get column name based on the string 'NASA (CRS)'.
# Note - Needs to disable regex cause of the brackets ()..
colName = ''
for col in df.columns:
    if df[col].astype(str).str.contains('NASA (CRS)', regex=False).any():
        colName = col

print("String Found in Columns: "+ colName)
# sum_nasa_crs = %sql SELECT SUM('PAYLOAD_MASS__KG_') AS TOTAL_PAYLOAD_MASS FROM SPACEXTABLE WHERE colName LIKE 'NASA (CRS)'
sum_qr =  "SELECT SUM(PAYLOAD_MASS__KG_) AS TOTAL_PAYLOAD_MASS FROM SPACEXTABLE WHERE Customer LIKE 'NASA (CRS)' "
sum_nasa_crs = pd.read_sql(sum_qr, con)
print(sum_nasa_crs)
```

```python
# Filter rows where Customer is 'NASA (CRS)' and sum the payload mass
total_mass = df[df['Customer'] == 'NASA (CRS)']['PAYLOAD_MASS__KG_'].sum()
print(f'Total Payload Mass carried by NASA (CRS): {total_mass} kg')
```

### Task 4




##### Display average payload mass carried by booster version F9 v1.1


```python
query = """
SELECT AVG(PAYLOAD_MASS__KG_) AS AVG_PAYLOAD_MASS
FROM SPACEXTABLE
WHERE Booster_Version = 'F9 v1.1'
"""

# Execute the query
avg_payload = pd.read_sql_query(query, con)
print(avg_payload)
```

```python
# Filter rows and calculate average payload mass
filtered_df = df[df['Booster_Version'] == 'F9 v1.1']
avg_payload_mass = filtered_df['PAYLOAD_MASS__KG_'].mean()

print(f"Average Payload Mass for Booster Version F9 v1.1: {avg_payload_mass} kg")
```

### Task 6

##### List the names of the boosters which have success in drone ship and have payload mass greater than 4000 but less than 6000


```python
query = """
SELECT DISTINCT Booster_Version
FROM SPACEXTABLE
WHERE Landing_Outcome = 'Success (drone ship)'
AND PAYLOAD_MASS__KG_ > 4000
AND PAYLOAD_MASS__KG_ < 6000
"""

result = pd.read_sql_query(query, con)
display(result)
```

### Task 7




##### List the total number of successful and failure mission outcomes

### Task 8



##### List all the booster_versions that have carried the maximum payload mass, using a subquery with a suitable aggregate function.


```python
# SQL Query to find booster versions with maximum payload mass
query = """
SELECT Booster_Version, PAYLOAD_MASS__KG_
FROM SPACEXTABLE
WHERE PAYLOAD_MASS__KG_ = (
    SELECT MAX(PAYLOAD_MASS__KG_)
    FROM SPACEXTABLE
)
"""

# Execute the query
result = pd.read_sql_query(query, con)
display(result)
```

```python
# Find the maximum payload mass
max_payload = df['PAYLOAD_MASS__KG_'].max()

# Filter rows where payload mass is equal to the maximum
max_payload_boosters = df[df['PAYLOAD_MASS__KG_'] == max_payload][['Booster_Version', 'PAYLOAD_MASS__KG_']]

display(max_payload_boosters)
```

### Task 9


##### List the records which will display the month names, failure landing_outcomes in drone ship ,booster versions, launch_site for the months in year 2015.

**Note: SQLLite does not support monthnames. So you need to use  substr(Date, 6,2) as month to get the months and substr(Date,0,5)='2015' for year.**


```python
qr = '''
SELECT 
    CASE substr(Date, 6, 2)
        WHEN '01' THEN 'January'
        WHEN '02' THEN 'February'
        WHEN '03' THEN 'March'
        WHEN '04' THEN 'April'
        WHEN '05' THEN 'May'
        WHEN '06' THEN 'June'
        WHEN '07' THEN 'July'
        WHEN '08' THEN 'August'
        WHEN '09' THEN 'September'
        WHEN '10' THEN 'October'
        WHEN '11' THEN 'November'
        WHEN '12' THEN 'December'
    END AS Month_Name,
    Landing_Outcome,
    Booster_Version,
    Launch_Site
FROM SPACEXTABLE
WHERE 
    Landing_Outcome = 'Failure (drone ship)'
    AND substr(Date, 0, 5) = '2015'
'''

res = pd.read_sql(qr, con)
display(res)
```

### Task 10




##### Rank the count of landing outcomes (such as Failure (drone ship) or Success (ground pad)) between the date 2010-06-04 and 2017-03-20, in descending order.


```python
qr = '''
SELECT 
    Landing_Outcome, 
    COUNT(*) AS Outcome_Count
FROM SPACEXTABLE
WHERE Date BETWEEN '2010-06-04' AND '2017-03-20'
GROUP BY Landing_Outcome
ORDER BY Outcome_Count DESC
'''

res = pd.read_sql(qr, con)
display(res)
```
